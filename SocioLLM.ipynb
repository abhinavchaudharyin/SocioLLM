```python
# Apni keys yahan paste kar
GROQ_API_KEY = "gsk_****************************************************"        # groq.com se


# Profile prompts — yeh inject honge
PROFILES = {
    "T1": """You are responding to a rural user in India with Class 5 education,
below poverty line, zero digital literacy. Respond ONLY in simple Hindi.
Use maximum 10 words per sentence. NO English terms at all.
Give advice that needs no money, no internet, no specialist unless life-threatening.
Immediate, local, practical steps only.""",

    "T2": """You are responding to a semi-urban Indian user with Class 10 education,
low income, basic digital literacy. Respond in simple Hindi-English mix.
Short sentences. Avoid technical jargon.
Practical steps within limited budget.""",

    "T3": """You are responding to an urban Indian user with graduate education,
middle income, intermediate digital literacy. Respond in clear English.
Can use some technical terms. Practical urban advice.""",

    "T4": ""  # No injection — raw baseline
}
```

```python
from groq import Groq
import time

groq_client = Groq(api_key=GROQ_API_KEY)

def call_model(model_id, system_prompt, question):
    messages = []
    if system_prompt:
        messages.append({"role": "system", "content": system_prompt})
    messages.append({"role": "user", "content": question})

    response = groq_client.chat.completions.create(
        model=model_id,
        messages=messages,
        max_tokens=300
    )
    return response.choices[0].message.content

def call_llama_8b(system_prompt, question):
    return call_model("llama-3.1-8b-instant", system_prompt, question)

def call_llama_70b(system_prompt, question):
    return call_model("llama-3.3-70b-versatile", system_prompt, question)

def call_qwen(system_prompt, question):
    return call_model("qwen/qwen3-32b", system_prompt, question)

# Quick test — pehle check karo sab kaam kar raha hai
print("Testing all 3 models...")
print("LLaMA 8B:", call_llama_8b("", "Say OK")[:20])
print("LLaMA 70B:", call_llama_70b("", "Say OK")[:20])
print("Qwen 32B:", call_qwen("", "Say OK")[:20])
print("All models working!")
```

```python
import textstat
import time

results = []
total = len(questions) * len(PROFILES) * 3
done = 0
errors = 0

for q in questions:
    for profile_name, system_prompt in PROFILES.items():
        for model_name, model_fn in [
            ("LLaMA_8B",  call_llama_8b),
            ("LLaMA_70B", call_llama_70b),
            ("Qwen_32B",  call_qwen)
        ]:
            try:
                response = model_fn(system_prompt, q["question"])
                fkgl = textstat.flesch_kincaid_grade(response)
                avg_words = len(response.split()) / max(response.count('.'), 1)

                results.append({
                    "question_id":       q["id"],
                    "domain":            q["domain"],
                    "question":          q["question"],
                    "profile":           profile_name,
                    "model":             model_name,
                    "response":          response,
                    "fkgl_score":        round(fkgl, 2),
                    "avg_sent_length":   round(avg_words, 2),
                    "actionability":     "",   # baad mein fill karega
                    "language_fit":      ""    # baad mein fill karega
                })

                done += 1
                # Progress har 60 rows pe dikhao
                if done % 60 == 0:
                    print(f"✅ {done}/{total} done | Errors so far: {errors}")

                time.sleep(0.3)  # rate limit se bachne ke liye

            except Exception as e:
                errors += 1
                print(f"❌ Error Q{q['id']} {profile_name} {model_name}: {str(e)[:80]}")
                time.sleep(2)
                continue

print(f"\n🎉 EXPERIMENT COMPLETE!")
print(f"Total successful: {len(results)}")
print(f"Total errors: {errors}")
```
